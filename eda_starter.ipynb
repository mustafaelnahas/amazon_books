BI-VARITE ANALYSIS

```python
df.head()
```

```python
df.groupby('Genre')['Price'].mean().sort_values(ascending=False).plot(kind='bar', figsize=(10, 5), color='orange')
```

```python
sns.scatterplot(x='Reviews', y='Price', data=df, hue='Genre', palette='Set1', alpha=0.7)
```

```python
sns.heatmap(df[[ 'Price' ,'Reviews']].corr(), annot=True)
```

```python
df.groupby('Year')['Reviews'].mean().plot(kind='line', figsize=(10, 5), color='purple')
```
